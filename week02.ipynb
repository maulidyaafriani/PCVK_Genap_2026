PERYANYAAN PRAKTIKUM

1.   Perbedaan hasil tampilan terjadi karena OpenCV menggunakan urutan kanal warna BGR, sedangkan Matplotlib menggunakan urutan RGB. Citra yang dibaca menggunakan cv2.imread() memiliki urutan kanal BGR sehingga ketika langsung ditampilkan menggunakan plt.imshow(), warna merah dan biru menjadi tertukar.
Nilai piksel pada array img tetap sama karena kedua cara tersebut menggunakan citra yang sama. Perbedaannya hanya pada cara kanal warna tersebut diinterpretasikan saat ditampilkan.
2.   Kanvas int16 dan int32 memiliki ukuran dimensi yang sama sehingga jumlah pikselnya sama. Nilai minimum dan maksimum pada kedua kanvas adalah 0 karena seluruh piksel diinisialisasi dengan nilai nol.Tampilan keduanya sama-sama berwarna hitam karena nilai pikselnya sama-sama 0. Namun ukuran memorinya berbeda. Tipe data int16 menggunakan 2 byte untuk setiap elemen, sedangkan int32 menggunakan 4 byte untuk setiap elemen. Oleh karena itu, kanvas int32 membutuhkan memori dua kali lebih besar dibandingkan int16.


1.   Baris from google.colab.patches import cv2_imshow digunakan untuk mengimpor fungsi cv2_imshow agar citra dapat ditampilkan dengan baik di Google Colab. Fungsi cv2.imshow() bawaan OpenCV tidak dapat digunakan secara normal di Google Colab karena cv2.imshow() membutuhkan tampilan window/GUI pada komputer lokal. Google Colab berjalan pada lingkungan notebook berbasis web sehingga tidak menyediakan window GUI seperti aplikasi desktop.
Oleh karena itu, cv2_imshow() digunakan sebagai fungsi untuk menampilkan citra langsung pada output notebook Colab.
2.   Citra yang dibaca menggunakan skimage.io.imread() sudah menggunakan urutan kanal RGB. Sementara itu, citra yang dibaca menggunakan cv2.imread() menggunakan urutan BGR sehingga perlu dikonversi menggunakan cv2.cvtColor(img, cv2.COLOR_BGR2RGB). Setelah dilakukan konversi BGR menjadi RGB, nilai piksel pada kedua citra akan memiliki urutan kanal yang sama. Oleh karena itu, skimage.io.imread() dan hasil cv2.cvtColor() sama-sama dapat menampilkan warna yang benar. Jika citra hasil cv2.imread() langsung ditampilkan menggunakan plt.imshow() tanpa konversi, kanal merah dan biru akan tertukar.



1.   Nilai img.shape tidak berubah setelah gambar ditampilkan dengan figsize yang lebih besar. Hal ini karena figsize hanya mengatur ukuran tampilan gambar pada Matplotlib, bukan mengubah jumlah piksel pada citra. Untuk benar-benar mengubah ukuran citra digunakan cv2.resize(). Jadi, cv2.resize() mengubah ukuran dan jumlah piksel citra, sedangkan figsize hanya mengubah ukuran tampilan.
