```python

# Dilasi tanpa library
def dilasi_citra(F, w):
    p, q = F.shape
    imgD = np.zeros((p, q), dtype=np.uint8)

    # Structure element dilasi (SED)
    SED = np.ones((w, w), dtype=np.uint8)
    c = (w - 1) // 2

    for i in range(c, p - c):
        for j in range(c, q - c):
            temp = F[i-c:i+c+1, j-c:j+c+1]
            product = temp * SED
            imgD[i, j] = np.max(product)
    return imgD


def proses_dilasi(path, w=5, thresh_val=125, inverse=False):
    # Cek file
    if not os.path.exists(path):
        print(f"File tidak ditemukan: {path}")
        return

    img = cv2.imread(path, 0)  # baca grayscale
    if img is None:
        print(f"Gagal membaca gambar: {path}")
        return

    # Binerisasi (inverse=True kalau objeknya hitam di latar putih)
    mode = cv2.THRESH_BINARY_INV if inverse else cv2.THRESH_BINARY
    _, thresh = cv2.threshold(img, thresh_val, 255, mode)

    imgD = dilasi_citra(thresh, w)

    plt.figure(figsize=(12, 4))
    plt.subplot(131), plt.imshow(img, cmap='gray')
    plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
    plt.subplot(132), plt.imshow(thresh, cmap='gray')
    plt.title('Hasil Threshold'), plt.xticks([]), plt.yticks([])
    plt.subplot(133), plt.imshow(imgD, cmap='gray')
    plt.title(f'Hasil Dilasi (w={w})'), plt.xticks([]), plt.yticks([])
    plt.show()


base = '/content/drive/MyDrive/PCVK/Images/244107020059/'

# Plat nomor
proses_dilasi(base + 'plat_nomer_biner.jpg', w=5)

# Noisy text
proses_dilasi(base + 'noisy_text.png', w=3, inverse=True)
```

```python
import cv2 as cv
from matplotlib import pyplot as plt
from skimage import filters
from skimage.morphology import skeletonize
# read image

img = cv.imread('/content/drive/MyDrive/PCVK/Images/244107020059/lily.jpg',0)
binary = img > filters.threshold_triangle(img)
# true false to one
binary_cp = binary.copy()
binary_cp[binary_cp == True] = 1
binary_cp[binary_cp == False] = 0
# skeletonize image
skeleton = skeletonize(binary_cp)
# print images
plt.figure(figsize=(20,20))
plt.subplot(121),plt.imshow(img,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(skeleton,cmap = 'gray')
plt.title('skeleton Image'), plt.xticks([]), plt.yticks([])
plt.savefig("binary and bin_skeleton.png")
plt.show()
```
