## Tugas 9 — Image Masking Privasi (KTM1d.jpg)

**Koordinat kotak mask yang dipakai:**
- Wajah : `(55, 293, 182, 462)` → (x1, y1, x2, y2)

**Tabel 5 Operator (hasil sesuai kode Anda — pakai `mask_3ch` sebagai operand kedua):**

| No. | Operator | Hasil |
|---|---|---|
| 1 | NOT (komplemen) | Seluruh citra jadi negatif (warna terbalik total), area mask (yang tadinya putih 255) menjadi hitam pada citra NOT karena `NOT(img)` tidak melibatkan mask sama sekali — hanya membalik nilai piksel citra asli. |
| 2 | OR (Atau) | Area mask (wajah & NIM) menjadi **putih solid**, karena `img OR mask_3ch` — bit manapun yang di-OR dengan 255 (semua bit 1) selalu menghasilkan 255. Area luar mask tetap seperti citra asli (karena OR dengan 0 = nilai itu sendiri). |
| 3 | AND (Dan) | Kebalikan dari OR: area **luar mask menjadi hitam total**, sedangkan area mask (wajah & NIM) tetap menampilkan citra asli — karena `img AND mask_3ch`, AND dengan 0 selalu 0, AND dengan 255 = nilai asli. Inilah prinsip yang dipakai (dengan `mask_inv`) di hasil masking utama Tugas 9. |
| 4 | NAND (Not AND) | Kebalikan logis dari AND: bagian yang sebelumnya "selamat" pada AND (wajah & NIM) kini jadi sangat gelap/kecil terlihat, sedangkan sisanya (area luar mask) berubah jadi warna terbalik (inverse) dari citra asli. |
| 5 | XOR (Exclusive Or) | Area di luar mask tetap seperti citra asli (XOR dengan 0 = nilai itu sendiri), sedangkan area mask (wajah & NIM) berubah warna jadi **komplemen/terbalik** (oranye/teal) karena XOR dengan 255 membalik nilai asli hanya di area tersebut. |

**Jawaban analisis:**

> Kelima operator logika memberikan efek berbeda tergantung bagaimana nilai bit mask (0 atau 255) berinteraksi dengan nilai bit citra asli. **NOT** independen dari mask, membalik seluruh citra. **OR** memaksa area mask jadi putih (sifat `1 OR x = 1`), sedangkan **AND** memaksa area di luar mask jadi hitam (sifat `0 AND x = 0`) — inilah dasar teknik penyamaran privasi: dengan memakai `mask_inv` (kebalikan mask), area sensitif (wajah & NIM) yang bernilai 0 pada mask_inv otomatis jadi hitam saat di-AND-kan. **NAND** adalah kebalikan logis dari AND sehingga polanya terbalik. **XOR** hanya mengubah nilai pada area yang overlap dengan mask (invert lokal), sementara area lain tidak tersentuh sama sekali — berbeda dari AND/OR yang memaksa salah satu sisi jadi konstan.
>
> Operator **AND** (dikombinasikan dengan `mask_inv`) adalah yang paling tepat untuk tujuan penyamaran data pribadi karena menghasilkan area hitam solid tanpa mengubah informasi di luar area sensitif, sedangkan operator lain lebih cocok untuk keperluan eksplorasi/visualisasi efek operasi logika saja.
